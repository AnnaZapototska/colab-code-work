# Task 1. Optimal product price

## Оптимальна ціна товару

Дано функцію попиту:

$$
Q(x) = 2000 - 0.8x
$$

де:
$$x$$ — ціна (грн)
$$Q(x)$$ — кількість продажів

Функція прибутку:

$$
P(x) = (x - 800)(2000 - 0.8x)
$$

---

### 1. Спрощення функції прибутку

Розкриємо дужки:

$$
P(x) = 2000x - 0.8x^2 - 1{,}600{,}000 + 640x
$$

$$
P(x) = -0.8x^2 + 2640x - 1{,}600{,}000
$$

---

### 2. Перша похідна

$$
P'(x) = -1.6x + 2640
$$

---

### 3. Критична точка

Знаходимо:

$$
P'(x) = 0
$$

$$
-1.6x + 2640 = 0
$$

$$
x^* = \frac{2640}{1.6} = 1650
$$

---

### 4. Друга похідна

$$
P''(x) = -1.6
$$

$$
P''(x) < 0
$$

Функція має максимум.

---

### 5. Максимальний прибуток

$$
P(1650) = (1650 - 800)(2000 - 0.8 \cdot 1650)
$$

$$
= 850 \cdot 680 = 578000
$$

---

### 6. Чисельна оптимізація

Використовуємо:

$$
\min(-P(x))
$$

на інтервалі:

$$
x \in [800, 2500]
$$

Отримуємо:

$$
x \approx 1650
$$

$$
P(x) \approx 578000
$$

---

### 7. Порівняння результатів

- Аналітичний результат:  
  $$x = 1650$$

- Чисельний результат:  
  $$x \approx 1650$$

✔️ Результати співпадають.

---

### Висновок

$$
x^* = 1650
$$

$$
P_{max} = 578000
$$

Оптимальна ціна становить **1650 грн**.
Це значення балансує між:
- зростанням ціни (прибуток за одиницю)
- зменшенням попиту

При цій ціні компанія отримує максимальний місячний прибуток.

```python
import numpy as np
from scipy.optimize import minimize_scalar

# Profit function
def profit(x):
    return (x - 800) * (2000 - 0.8 * x)

#  1. Analytical solution
print(" Pricing Optimization")
print("-" * 40)

x_analytical = 2640 / 1.6

# Second derivative:
second_derivative = -1.6

p_analytical = profit(x_analytical)

print("\n1. Analytical solution:")
print(f"P'(x) = -1.6x + 2640")
print(f"Critical point x* = {x_analytical:.2f} UAH")

print("\n2. Second derivative test:")
print(f"P''(x) = {second_derivative}")
if second_derivative < 0:
    print("P''(x) < 0 -> Maximum")
else:
    print("Minimum")

print(f"\nMaximum profit (analytical) = {p_analytical:.2f} UAH")

# 2. Numerical solution
def negative_profit(x):
    return -profit(x)

result = minimize_scalar(negative_profit, bounds=(800, 2500), method='bounded')

x_num = result.x
p_num = profit(x_num)

print("\n3. Numerical optimization:")
print(f"Optimal price = {x_num:.2f} UAH")
print(f"Max profit = {p_num:.2f} UAH")

# 3. Comparison
print("\n4. Comparison:")
print(f"Price difference = {abs(x_num - x_analytical):.4f}")
print(f"Profit difference = {abs(p_num - p_analytical):.4f}")

# 4. Final conclusion
print("\n5. Final result:")
print(f"Optimal price = {x_analytical:.2f} UAH")
print(f"Maximum monthly profit = {p_analytical:.2f} UAH")
```

# Task 2. Minimizing a function of two variables

## Мінімізація функції двох змінних

Розглянемо функцію:

$$
f(x, y) = x^2 + xy + y^2 - 6x - 9y + 20
$$

---

### 1. Критична точка

Обчислюємо частинні похідні:

$$
\frac{\partial f}{\partial x} = 2x + y - 6
$$

$$
\frac{\partial f}{\partial y} = x + 2y - 9
$$

Прирівнюємо до нуля:

$$
\begin{cases}
2x + y - 6 = 0 \\
x + 2y - 9 = 0
\end{cases}
$$

Розв’язуємо систему:

$$
y = 6 - 2x
$$

$$
x + 2(6 - 2x) = 9
$$

$$
x + 12 - 4x = 9
$$

$$
-3x = -3 \Rightarrow x = 1
$$

$$
y = 6 - 2 = 4
$$

---

### Критична точка

$$
(x^*, y^*) = (1, 4)
$$

---

### 2. Матриця Гессе

Другі похідні:

$$
\frac{\partial^2 f}{\partial x^2} = 2
$$

$$
\frac{\partial^2 f}{\partial y^2} = 2
$$

$$
\frac{\partial^2 f}{\partial x \partial y} = 1
$$

Матриця Гессе:

$$
H =
\begin{pmatrix}
2 & 1 \\
1 & 2
\end{pmatrix}
$$

---

### 3. Характеристичне рівняння

$$
\det(H - \lambda I) = 0
$$

$$
\begin{vmatrix}
2 - \lambda & 1 \\
1 & 2 - \lambda
\end{vmatrix} = 0
$$

$$
(2 - \lambda)^2 - 1 = 0
$$

$$
(2 - \lambda)^2 = 1
$$

$$
\lambda_1 = 1,\quad \lambda_2 = 3
$$

---

### 4. Висновок про тип екстремуму

$$
\lambda_1 > 0,\quad \lambda_2 > 0
$$

Матриця Гессе додатно визначена.
Точка є **локальним мінімумом**.

---

### 5. Чисельна оптимізація (BFGS)

Запускаємо метод BFGS з початкової точки:

$$
(0, 0)
$$

Отримуємо:

$$
(x, y) \approx (1, 4)
$$

$$
f(x, y) = -1
$$

---

### 6. Перевірка стійкості

Запускаємо оптимізацію з різних точок:

- $$(0, 0)$$  
- $$(10, 10)$$  
- $$(-5, 15)$$  

У всіх випадках:

$$
(x, y) \to (1, 4)
$$

---

### Висновок

$$
(x^*, y^*) = (1, 4)
$$

$$
f_{min} = -1
$$

Функція має єдиний глобальний мінімум.
Результат підтверджено:
- аналітично (через похідні та Гессе)
- чисельно (метод BFGS)
- перевіркою стійкості

Мінімум стабільний і не залежить від початкової точки.

```python
import numpy as np
import sympy as sp
from scipy.optimize import minimize

print("Optimization of Function of Two Variables")
print("-" * 50)

# 1. Analytical solution
x, y = sp.symbols('x y')

f_sym = x**2 + x*y + y**2 - 6*x - 9*y + 20

# partial derivatives
fx = sp.diff(f_sym, x)
fy = sp.diff(f_sym, y)

print("\n1. Partial derivatives:")
print(f"∂f/∂x = {fx}")
print(f"∂f/∂y = {fy}")

# solve system
solution = sp.solve([fx, fy], (x, y))
x_star, y_star = solution[x], solution[y]

print("\nCritical point:")
print(f"(x*, y*) = ({x_star}, {y_star})")

# 2. Hessian and type of extremum
H = sp.hessian(f_sym, (x, y))

print("\n2. Hessian matrix:")
sp.pprint(H)

# characteristic equation
lam = sp.symbols('lambda')
char_eq = sp.det(H - lam * sp.eye(2))

print("\nCharacteristic equation:")
sp.pprint(char_eq)

eigenvalues = H.eigenvals()

print("\nEigenvalues:")
print(eigenvalues)

print("\nConclusion:")
print("Both eigenvalues > 0 -> Hessian is positive definite -> minimum")

# 3. Numerical optimization
def f(v):
    x, y = v
    return x**2 + x*y + y**2 - 6*x - 9*y + 20

initial_guess = [0, 0]

result = minimize(f, initial_guess, method='BFGS')

x_opt, y_opt = result.x

print("\n3. Numerical optimization (BFGS):")
print(f"Optimal point = ({x_opt:.2f}, {y_opt:.2f})")
print(f"Minimum value = {result.fun:.2f}")

# 4. Robustness check
initial_points = [
    (0, 0),
    (10, 10),
    (-5, 15)
]

print("\n4. Robustness check:")
results = []

for point in initial_points:
    res = minimize(f, point, method='BFGS')
    x_r, y_r = res.x
    results.append((round(x_r, 4), round(y_r, 4)))

    print(f"\nStart point: {point}")
    print(f"-> Converged to: ({x_r:.4f}, {y_r:.4f})")
    print(f"-> Function value: {res.fun:.4f}")
    print(f"-> Success: {res.success}")

print("\nConsistency check:")
for r in results:
    print(r)

print("\nFinal conclusion:")
print("All runs converge to the same point -> global minimum confirmed")
```

# Task 3. Optimal production plan

## Оптимальний план виробництва

Меблева фабрика виготовляє:

$$x$$ кількість стільців $$y$$ кількість столів  

---

### 1. Цільова функція

Прибуток:

$$
P(x,y) = 500x + 800y
$$

Мета:

$$
\max P(x,y)
$$

---

### 2. Обмеження

#### Деревина:

$$
2x + 4y \le 120
$$

#### Робочий час:

$$
3x + 2y \le 90
$$

#### Невід’ємність:

$$
x \ge 0,\quad y \ge 0
$$

---

### 3. Математична модель

$$
\max P(x,y) = 500x + 800y
$$

$$
\begin{cases}
2x + 4y \le 120 \\
3x + 2y \le 90 \\
x \ge 0,\ y \ge 0
\end{cases}
$$

---

### 4. Аналітичне рішення

Знаходимо вершини області допустимих рішень.

Перетин обмежень:

$$
\begin{cases}
2x + 4y = 120 \\
3x + 2y = 90
\end{cases}
$$

Спрощуємо:

$$
x + 2y = 60
$$

$$
x = 60 - 2y
$$

Підставляємо:

$$
3(60 - 2y) + 2y = 90
$$

$$
180 - 6y + 2y = 90
$$

$$
-4y = -90 \Rightarrow y = 22.5
$$

$$
x = 60 - 45 = 15
$$

---

### 5. Перевірка вершин

| Точка | Прибуток |
|------|--------|
| (0,0) | 0 |
| (30,0) | 15000 |
| (0,30) | 24000 |
| (15,22.5) | **25500** |

---

### Оптимальне рішення

$$
x^* = 15,\quad y^* = 22.5
$$

$$
P_{max} = 25500
$$

---

### 6. Чисельне рішення (linprog)

Мінімізуємо:

$$
- P(x,y)
$$

Отримуємо:

$$
x \approx 15,\quad y \approx 22.5
$$

$$
P \approx 25500
$$

✔️ Результат співпадає з аналітичним.

---

### 7. Аналіз ресурсів

Використання деревини:

$$
2x + 4y = 120
$$

Використання часу:

$$
3x + 2y = 90
$$

---

### Висновок

$$
(x^*, y^*) = (15, 22.5)
$$

$$
P_{max} = 25500
$$

Обидва ресурси використані повністю. Це оптимальний баланс між виробництвом стільців і столів. Будь-яке інше рішення дає менший прибуток.

```python
from scipy.optimize import linprog

# coefficients for minimization (-profit)
c = [-500, -800]

# constraints
A = [
    [2, 4],
    [3, 2]
]

b = [120, 90]

# bounds
x_bounds = (0, None)
y_bounds = (0, None)

result = linprog(c, A_ub=A, b_ub=b, bounds=[x_bounds, y_bounds], method='highs')

x_opt, y_opt = result.x

profit = 500 * x_opt + 800 * y_opt

print("Optimal Production Plan")
print("-" * 40)

print(f"Chairs = {x_opt:.0f}")
print(f"Tables = {y_opt:.0f}")
print(f"Max profit = {profit:.2f} UAH")

# Resource usage
wood_used = 2*x_opt + 4*y_opt
time_used = 3*x_opt + 2*y_opt

print("\nResource usage:")
print(f"Wood used = {wood_used:.2f} / 120")
print(f"Time used = {time_used:.2f} / 90")

print("\nRemaining:")
print(f"Wood left = {120 - wood_used:.2f}")
print(f"Time left = {90 - time_used:.2f}")
```

# Task 4. Comprehensive production analysis

## Комплексний аналіз виробництва

Завод виробляє продукцію протягом 10-годинної зміни.

---

### 1. Продуктивність

Продуктивність (одиниць на годину) задається функцією:

$$
P(t) = 100 + 40t - 4t^2
$$

де:
- $$t$$ — час (години), $$0 \le t \le 10$$

---

### 2. Дослідження швидкості зміни

Похідна продуктивності:

$$
P'(t) = 40 - 8t
$$

---

#### Аналіз у точках:

- $$t = 2$$:
$$
P'(2) = 40 - 16 = 24 > 0
$$
продуктивність **зростає**

- $$t = 5$$:
$$
P'(5) = 40 - 40 = 0
$$
**максимум (стаціонарна точка)**

- $$t = 8$$:
$$
P'(8) = 40 - 64 = -24 < 0
$$
продуктивність **спадає**

---

### 3. Максимальна продуктивність

Знаходимо максимум:

$$
P'(t) = 0
$$

$$
40 - 8t = 0 \Rightarrow t^* = 5
$$

---

Обчислюємо значення:

$$
P(5) = 100 + 200 - 100 = 200
$$

---

### Пік продуктивності:

$$
t^* = 5 \text{ год}
$$

$$
P_{max} = 200 \text{ од/год}
$$

---

### 4. Загальний обсяг виробництва

Обчислюємо інтеграл:

$$
\int_0^{10} P(t)\,dt
$$

---

Аналітичне рішення:

$$
\int (100 + 40t - 4t^2) dt =
100t + 20t^2 - \frac{4}{3}t^3
$$

---

Підстановка:

$$
\left[100t + 20t^2 - \frac{4}{3}t^3\right]_0^{10}
$$

$$
= 1000 + 2000 - \frac{4000}{3}
$$

$$
= 3000 - 1333.33 = 1666.67
$$

---

### Загальний випуск:

$$
Q = 1666.67 \text{ одиниць}
$$

---

## Частина 2. Оптимізація витрат

---

### 5. Функція вартості

$$
C(x,y) = x^2 + y^2 - 10x - 8y + 50
$$

де:
- $$x$$ — витрати на сировину  
- $$y$$ — витрати на енергію  

---

### 6. Початкові параметри

Система:

$$
\begin{cases}
2x + y = 20 \\
x + 3y = 25
\end{cases}
$$

---

Розв'язання:

$$
y = 20 - 2x
$$

$$
x + 3(20 - 2x) = 25
$$

$$
x + 60 - 6x = 25
$$

$$
-5x = -35 \Rightarrow x = 7
$$

$$
y = 6
$$

---

### Початкова точка:

$$
(x_0, y_0) = (7, 6)
$$

---

### 7. Мінімізація вартості

Знаходимо частинні похідні:

$$
\frac{\partial C}{\partial x} = 2x - 10
$$

$$
\frac{\partial C}{\partial y} = 2y - 8
$$

---

Прирівнюємо до нуля:

$$
2x - 10 = 0 \Rightarrow x^* = 5
$$

$$
2y - 8 = 0 \Rightarrow y^* = 4
$$

---

### Мінімальна вартість:

$$
C(5,4) = 25 + 16 - 50 - 32 + 50 = 9
$$

---

## Фінальний результат

---

### Загальна вартість виробництва:

$$
\text{Total Cost} = Q \cdot C_{min}
$$

$$
= 1666.67 \cdot 9
$$

$$
= 15000
$$

---

## Висновок

- Продуктивність досягає максимуму в середині зміни:
$$
t^* = 5
$$

- Максимальна продуктивність:
$$
P_{max} = 200
$$

- Загальний обсяг виробництва:
$$
Q = 1666.67
$$

- Оптимальні витрати:
$$
(x^*, y^*) = (5, 4)
$$

- Мінімальна вартість одиниці:
$$
C_{min} = 9
$$

---

### Остаточний результат:

$$
\text{Total Cost} = 15000
$$

```python
import numpy as np
from scipy.optimize import approx_fprime, minimize_scalar, minimize
from scipy.integrate import quad

# Productivity function P(t)
def productivity(t):
    return 100 + 40*t - 4*t**2

print("--- PART 1: PRODUCTIVITY ANALYSIS ---\n")

# 1. Derivative using approx_fprime
for t in [2, 5, 8]:
    grad = approx_fprime(np.array([t]), lambda x: productivity(x[0]), epsilon = 1e-8)
    derivative = grad[0]
    norm = np.linalg.norm(grad)

    if derivative > 0:
        trend = "increasing"
    elif derivative < 0:
        trend = "decreasing"
    else:
        trend = "stationary"

    print(f"t = {t} h: P'(t) = {derivative:+.2f} (trend: {trend}), ||grad|| = {norm:.2f}")

print()

# 2. Find peak productivity
result_max = minimize_scalar(lambda t: -productivity(t), bounds=(0, 10), method='bounded')
t_star = result_max.x
P_max = productivity(t_star)

print(f"Peak productivity time t* = {t_star:.2f} h")
print(f"Maximum productivity P(t*) = {P_max:.2f} units/hour\n")

# 3. Total production over the shift
total_production, _ = quad(productivity, 0, 10)

print(f"Total production over 10 hours = {total_production:.2f} units\n")

print("--- PART 2: COST OPTIMIZATION --- \n")

def cost(vars):
    x, y = vars
    return x**2 + y**2 - 10*x - 8*y + 50

# 4. Solve system of equations to find initial point
A = np.array([[2, 1],
              [1, 3]])
b = np.array([20, 25])

x0, y0 = np.linalg.solve(A, b)

print(f"Initial parameters from system: x0 = {x0:.2f}, y0 = {y0:.2f}")

# 5. Minimize cost using BFGS
result_min = minimize(cost, x0=[x0, y0], method='BFGS')

x_opt, y_opt = result_min.x
C_min = result_min.fun

print(f"Optimal parameters: x* = {x_opt:.2f}, y* = {y_opt:.2f}")
print(f"Minimum cost per unit = {C_min:.2f}\n")

# Final result

total_cost = total_production * C_min

print("--- FINAL RESULT ---\n")
print(f"Total production = {total_production:.2f} units")
print(f"Minimum cost per unit = {C_min:.2f}")
print(f"Total production cost = {total_cost:.2f}")
```
