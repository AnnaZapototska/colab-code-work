# Task 1. Mobile app ratings analysis

## Аналіз рейтингів мобільного додатку

Дано розподіл оцінок користувачів:

$$
X \in \{1,2,3,4,5\}
$$

Кількість відгуків:

$$
\{120, 180, 320, 580, 800\}
$$

Загальна кількість:

$$
N = 2000
$$

---

### 1. Функція маси ймовірності (PMF)

$$
P(X = x) = \frac{n_x}{N}
$$

Обчислення:

$$
P(1) = \frac{120}{2000} = 0.06
$$

$$
P(2) = \frac{180}{2000} = 0.09
$$

$$
P(3) = \frac{320}{2000} = 0.16
$$

$$
P(4) = \frac{580}{2000} = 0.29
$$

$$
P(5) = \frac{800}{2000} = 0.40
$$

---

### 2. Перевірка нормування

$$
\sum P(X=x) = 0.06 + 0.09 + 0.16 + 0.29 + 0.40 = 1.00
$$

✔️ Розподіл коректний.

---

### 3. Функція розподілу (CDF)

$$
F(x) = P(X \le x)
$$

$$
F(1) = 0.06
$$

$$
F(2) = 0.06 + 0.09 = 0.15
$$

$$
F(3) = 0.15 + 0.16 = 0.31
$$

$$
F(4) = 0.31 + 0.29 = 0.60
$$

$$
F(5) = 1.00
$$

---

### 4. Математичне сподівання

$$
E[X] = \sum x \cdot P(X=x)
$$

$$
E[X] = 1\cdot0.06 + 2\cdot0.09 + 3\cdot0.16 + 4\cdot0.29 + 5\cdot0.40
$$

$$
E[X] = 3.88
$$

---

### 5. Дисперсія

$$
E[X^2] = \sum x^2 \cdot P(X=x)
$$

$$
E[X^2] = 16.50
$$

$$
Var(X) = E[X^2] - (E[X])^2
$$

$$
Var(X) = 16.50 - (3.88)^2 \approx 1.45
$$

---

### 6. Стандартне відхилення

$$
\sigma = \sqrt{Var(X)}
$$

$$
\sigma \approx 1.20
$$

---

### 7. Ймовірність негативної оцінки

$$
P(X \le 2) = P(1) + P(2)
$$

$$
P(X \le 2) = 0.06 + 0.09 = 0.15
$$

---

### 8. Ймовірність позитивної оцінки

$$
P(X \ge 4) = P(4) + P(5)
$$

$$
P(X \ge 4) = 0.29 + 0.40 = 0.69
$$

---

### 9. Медіана

$$
F(x) \ge 0.5
$$

$$
F(3) = 0.31 < 0.5
$$

$$
F(4) = 0.60 \ge 0.5
$$

$$
\text{Median} = 4
$$

---

### Висновок

$$
E[X] = 3.88
$$

$$
\sigma \approx 1.20
$$

Більшість користувачів ставлять високі оцінки:

$$
P(X \ge 4) = 0.69
$$

Негативних оцінок небагато:

$$
P(X \le 2) = 0.15
$$

Додаток має позитивне сприйняття користувачами.

```python
import pandas as pd
import numpy as np

# data
ratings = {
    1: 120,
    2: 180,
    3: 320,
    4: 580,
    5: 800
}

df = pd.DataFrame(list(ratings.items()), columns=["Rating", "Number"])
df = df.sort_values("Rating")

# PMF and CDF
total = df["Number"].sum()
df["PMF"] = df["Number"] / total
df["CDF"] = df["PMF"].cumsum()

df_display = df.copy()
df_display["PMF"] = df_display["PMF"].map("{:.2f}".format)
df_display["CDF"] = df_display["CDF"].map("{:.2f}".format)

print("\nApp rating distribution:\n")
print(df_display[["Rating", "PMF", "CDF"]].to_string(index=False))

x = df["Rating"].values
p = df["PMF"].values

# check PMF
pmf_sum = df["PMF"].sum()
print(f"\nCheck: sum of PMF = {pmf_sum:.2f}")

if np.isclose(pmf_sum, 1.0):
    print("PMF is valid (sum = 1)")
else:
    print("PMF is NOT valid")

# statistics
E = np.sum(x * p)
E2 = np.sum((x**2) * p)
Var = E2 - E**2
std = np.sqrt(Var)

print(f"\nE[X] = {E:.2f}")
print(f"Var(X) = {Var:.2f}")
print(f"σ = {std:.2f}")

prob_negative = p[x <= 2].sum()
prob_positive = p[x >= 4].sum()

print(f"\nP(Negative rating 1 or 2) = {prob_negative:.2f}")
print(f"P(X ≥ 4) = {prob_positive:.2f}")

# median
cdf = df["CDF"].values
median = x[cdf >= 0.5][0]

print(f"\nMedian = {median}")
```

# Task 2. Order delivery analysis

## Аналіз роботи служби доставки

Кількість замовлень:

$$
n = 50
$$

Ймовірність вчасної доставки:

$$
p = 0.88
$$

Випадкова величина:

$$
X \sim Binomial(n=50, p=0.88)
$$

---

### 1. Математичне сподівання

$$
E[X] = n \cdot p
$$

$$
E[X] = 50 \cdot 0.88 = 44
$$

---

### 2. Стандартне відхилення

$$
\sigma = \sqrt{n \cdot p \cdot (1 - p)}
$$

$$
\sigma = \sqrt{50 \cdot 0.88 \cdot 0.12}
$$

$$
\sigma = \sqrt{5.28} \approx 2.30
$$

---

### 3. Ймовірність, що всі замовлення доставлять вчасно

$$
P(X = 50) = \binom{50}{50} \cdot 0.88^{50} \cdot 0.12^0
$$

$$
P(X = 50) = 0.88^{50} \approx 0.00166
$$

---

### 4. Ймовірність, що рівно 45 замовлень доставлять вчасно

$$
P(X = 45) = \binom{50}{45} \cdot 0.88^{45} \cdot 0.12^{5}
$$

$$
P(X = 45) \approx 0.152
$$

---

### 5. Ймовірність, що вчасно доставлять від 42 до 46 замовлень

$$
P(42 \le X \le 46) = \sum_{k=42}^{46} P(X = k)
$$

Або через функцію розподілу:

$$
P(42 \le X \le 46) = F(46) - F(41)
$$

$$
P(42 \le X \le 46) \approx 0.73
$$

---

### 6. Ймовірність, що більше 5 замовлень запізняться

Запізнені замовлення:

$$
50 - X
$$

Умова:

$$
50 - X > 5
$$

$$
X < 45
$$

Отже:

$$
P(X < 45) = P(X \le 44)
$$

$$
P(X \le 44) = F(44) \approx 0.45
$$

---

### Висновок

$$
E[X] = 44
$$

$$
\sigma \approx 2.30
$$

Типова кількість вчасних доставок — близько 44.

Ймовірність ідеального дня дуже мала:

$$
P(X = 50) \approx 0.00166
$$

Найбільш імовірний діапазон:

$$
P(42 \le X \le 46) \approx 0.73
$$

Ймовірність, що більше 5 замовлень запізняться:

$$
P(X \le 44) \approx 0.45
$$

```python
from scipy.stats import binom

# parameters
n = 50
p = 0.88

print("Delivery Service Analysis")
# 1. Expected value and standard deviation
E = binom.mean(n, p)
std = binom.std(n, p)

print("\n1. Expected number of on-time deliveries:")
print(f"E[X] = {E:.2f}")

print("\nStandard deviation:")
print(f"σ = {std:.2f}")

# 2. Probability all deliveries are on time
prob_all = binom.pmf(50, n, p)

print("\n2. Probability that all 50 deliveries are on time:")
print(f"P(X = 50) = {prob_all:.5f}")

# 3. Probability exactly 45 deliveries are on time
prob_45 = binom.pmf(45, n, p)

print("\n3. Probability that exactly 45 deliveries are on time:")
print(f"P(X = 45) = {prob_45:.5f}")

# 4. Probability between 42 and 46 deliveries
prob_range = binom.cdf(46, n, p) - binom.cdf(41, n, p)

print("\n4. Probability that between 42 and 46 deliveries are on time:")
print(f"P(42 ≤ X ≤ 46) = {prob_range:.4f}")

# 5. Probability that more than 5 deliveries are late
prob_late = binom.cdf(44, n, p)

print("\n5. Probability that more than 5 deliveries are late:")
print(f"P(X ≤ 44) = {prob_late:.4f}")
```

# Task 3. Time between orders in the online store

## Час між замовленнями в інтернет-магазині

Дано:

Інтенсивність потоку замовлень:

$$
\lambda = 6 \text{ замовлень/год} = 0.1 \text{ замовлень/хв}
$$

Випадкова величина:

$$
T \sim Exponential(\lambda = 0.1)
$$

Середній час між подіями:

$$
\text{scale} = \frac{1}{\lambda} = 10 \text{ хвилин}
$$

---

### 1. Математичне сподівання

$$
E[T] = \frac{1}{\lambda}
$$

$$
E[T] = \frac{1}{0.1} = 10
$$

---

### 2. Стандартне відхилення

$$
\sigma = \frac{1}{\lambda}
$$

$$
\sigma = 10
$$

---

### 3. Ймовірність, що замовлення надійде менше ніж за 5 хвилин

$$
P(T < t) = 1 - e^{-\lambda t}
$$

$$
P(T < 5) = 1 - e^{-0.1 \cdot 5}
$$

$$
P(T < 5) = 1 - e^{-0.5} \approx 0.3935
$$

---

### 4. Ймовірність, що доведеться чекати більше 15 хвилин

$$
P(T > t) = e^{-\lambda t}
$$

$$
P(T > 15) = e^{-0.1 \cdot 15}
$$

$$
P(T > 15) = e^{-1.5} \approx 0.2231
$$

---

### 5. Ймовірність, що замовлення надійде в проміжку від 5 до 15 хвилин

$$
P(a < T < b) = F(b) - F(a)
$$

$$
F(t) = 1 - e^{-\lambda t}
$$

$$
P(5 < T < 15) = e^{-0.5} - e^{-1.5}
$$

$$
P(5 < T < 15) \approx 0.3834
$$

---

### 6. Медіанний час очікування

Медіана визначається як:

$$
F(t) = 0.5
$$

$$
1 - e^{-\lambda t} = 0.5
$$

$$
e^{-\lambda t} = 0.5
$$

$$
t = \frac{\ln(2)}{\lambda}
$$

$$
t = \frac{0.693}{0.1} \approx 6.93
$$

---

### Висновок

$$
E[T] = 10
$$

$$
\sigma = 10
$$

Середній час між замовленнями — 10 хвилин.

Ймовірності:

$$
P(T < 5) \approx 0.39
$$

$$
P(T > 15) \approx 0.22
$$

$$
P(5 < T < 15) \approx 0.38
$$

Медіанний час:

$$
T_{median} \approx 6.93 \text{ хв}
$$

Потік замовлень нерівномірний: багато замовлень приходять швидко, але іноді доводиться чекати значно довше.

```python
from scipy.stats import expon

# Parameters
lambda_rate = 0.1
scale = 1 / lambda_rate

print("Online Store Order Flow Analysis")
print("-" * 40)
# 1. Expected value and standard deviation
E = expon.mean(scale=scale)
std = expon.std(scale=scale)

print("\n1. Expected time between orders:")
print(f"E[T] = {E:.2f} minutes")

print("\n2. Standard deviation:")
print(f"σ = {std:.2f} minutes")

# 2. Probability: less than 5 minutes
prob_less_5 = expon.cdf(5, scale=scale)

print("\n3. Probability that the next order arrives in less than 5 minutes:")
print(f"P(T < 5) = {prob_less_5:.4f} ({prob_less_5*100:.2f}%)")

# 3. Probability: more than 15 minutes
prob_more_15 = expon.sf(15, scale=scale)

print("\n4. Probability that waiting time exceeds 15 minutes:")
print(f"P(T > 15) = {prob_more_15:.4f} ({prob_more_15*100:.2f}%)")

# 4. Probability: between 5 and 15 minutes
prob_5_15 = expon.cdf(15, scale=scale) - expon.cdf(5, scale=scale)

print("\n5. Probability that the next order arrives between 5 and 15 minutes:")
print(f"P(5 < T < 15) = {prob_5_15:.4f} ({prob_5_15*100:.2f}%)")

# 5. Median
median = expon.ppf(0.5, scale=scale)

print("\n6. Median waiting time:")
print(f"Median(T) = {median:.2f} minutes")

print("\nInterpretation:")
print("- On average, an order arrives every 10 minutes")
print("- About 39% of orders arrive within 5 minutes")
print("- About 22% of cases require waiting more than 15 minutes")
print("- The most typical waiting time (median) is ~7 minutes")
```

# Task 4. Checking students' GPA

## Перевірка середнього балу студентів

Дано:

Історичне середнє:

$$
\mu_0 = 72
$$

Вибіркові дані:

$$
\bar{x} = 76.8,\quad s = 14.2,\quad n = 35
$$

---

### 1. Формулювання гіпотез

Нульова гіпотеза:

$$
H_0: \mu = 72
$$

Альтернативна гіпотеза:

$$
H_1: \mu \ne 72
$$

Двосторонній тест.

---

### 2. Стандартна похибка

$$
SE = \frac{s}{\sqrt{n}}
$$

$$
SE = \frac{14.2}{\sqrt{35}} \approx 2.40
$$

---

### 3. t-статистика

$$
t = \frac{\bar{x} - \mu_0}{SE}
$$

$$
t = \frac{76.8 - 72}{2.40} = 2.00
$$

---

### 4. p-value

Для двостороннього тесту:

$$
p = 2 \cdot (1 - F(|t|))
$$

$$
p \approx 0.052
$$

---

### 5. 95% довірчий інтервал

$$
CI = \bar{x} \pm t_{crit} \cdot SE
$$

Ступені свободи:

$$
df = n - 1 = 34
$$

Критичне значення:

$$
t_{crit} \approx 2.03
$$

Похибка:

$$
ME = 2.03 \cdot 2.40 \approx 4.87
$$

Інтервал:

$$
(76.8 - 4.87,\; 76.8 + 4.87)
$$

$$
CI \approx (71.93,\; 81.67)
$$

---

### 6. Висновок

Рівень значущості:

$$
\alpha = 0.05
$$

Порівняння:

$$
p = 0.052 > 0.05
$$

Нульова гіпотеза **не відхиляється**.

---

### Інтерпретація

- Недостатньо статистичних доказів, що нова методика вплинула на результати
- Результат знаходиться на межі значущості
- Історичне середнє входить у довірчий інтервал:

$$
72 \in (71.93,\; 81.67)
$$

Для підтвердження еффекту потрібна більша вибірка.

```python
import numpy as np
from scipy.stats import t

# Data
mu_0 = 72
x_bar = 76.8
s = 14.2
n = 35
alpha = 0.05

# 1. Hypotheses
print("Student Score Analysis")
print("-" * 40)

print("1. Hypotheses:")
print("H0: μ = 72")
print("H1: μ ≠ 72 (two-sided test)")

# 2. Standard Error
SE = s / np.sqrt(n)

print("\n2. Standard Error:")
print(f"SE = {SE:.2f}")

# 3. t-statistic
t_stat = (x_bar - mu_0) / SE

print("\n3. t-statistic:")
print(f"t = {t_stat:.2f}")

# 4. p-value
df = n - 1
p_value = 2 * (1 - t.cdf(abs(t_stat), df))

print("\n4. p-value (two-sided):")
print(f"p-value = {p_value:.4f}")

# 5. Confidence Interval
t_crit = t.ppf(1 - alpha/2, df)
ME = t_crit * SE

lower = x_bar - ME
upper = x_bar + ME

print("\n5. 95% Confidence Interval:")
print(f"t_crit = {t_crit:.2f}")
print(f"CI = ({lower:.2f}, {upper:.2f})")

# 6. Conclusion
print("\n6. Conclusion:")

if p_value <= alpha:
    print("Reject H0")
    print("There is evidence that the new method affected the results.")
else:
    print("Fail to reject H0")
    print("There is not enough evidence to conclude that the new method affected the results.")
```
